```python

# Resolution in First Order Logic
# Prove: John likes peanuts

from itertools import combinations

# Knowledge Base in CNF

clauses = [
    # a. John likes all kinds of food
    ["~Food(x)", "Likes(John,x)"],

    # b. Apple and vegetables are food
    ["Food(Apple)"],
    ["Food(Vegetables)"],

    # c. Anything anyone eats and is not killed is food
    ["~Eats(x,y)", "~NotKilled(x)", "Food(y)"],

    # d. Anil eats peanuts and is still alive
    ["Eats(Anil,Peanuts)"],
    ["Alive(Anil)"],

    # e. Harry eats everything Anil eats
    ["~Eats(Anil,x)", "Eats(Harry,x)"],

    # f. Anyone alive implies not killed
    ["~Alive(x)", "NotKilled(x)"],

    # g. Anyone not killed implies alive
    ["~NotKilled(x)", "Alive(x)"]
]

# Negation of goal:
# Goal: Likes(John,Peanuts)
# Negated goal: ~Likes(John,Peanuts)

clauses.append(["~Likes(John,Peanuts)"])


# Unification function
def unify(literal1, literal2):
    if literal1[1:] == literal2[1:]:
        return True
    return False


# Resolution function
def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:

            if l1.startswith("~") and not l2.startswith("~"):
                if unify(l1, l2):
                    new_clause = [
                        x for x in c1 if x != l1
                    ] + [
                        x for x in c2 if x != l2
                    ]
                    return new_clause

            elif not l1.startswith("~") and l2.startswith("~"):
                if unify(l1, l2):
                    new_clause = [
                        x for x in c1 if x != l1
                    ] + [
                        x for x in c2 if x != l2
                    ]
                    return new_clause

    return None


# Apply resolution
print("Initial Clauses:")
for i, clause in enumerate(clauses):
    print(f"C{i+1}: {' OR '.join(clause)}")

print("\nResolution Steps:")

# Direct resolution demonstration
# From Alive(Anil) and rule f:
# Derive NotKilled(Anil)

print("\nStep 1:")
print("Alive(Anil)")
print("~Alive(x) OR NotKilled(x)")
print("Therefore: NotKilled(Anil)")

# From Eats(Anil,Peanuts) and rule c:
# Derive Food(Peanuts)

print("\nStep 2:")
print("Eats(Anil,Peanuts)")
print("NotKilled(Anil)")
print("~Eats(x,y) OR ~NotKilled(x) OR Food(y)")
print("Therefore: Food(Peanuts)")

# From Food(Peanuts) and rule a:
# Derive Likes(John,Peanuts)

print("\nStep 3:")
print("Food(Peanuts)")
print("~Food(x) OR Likes(John,x)")
print("Therefore: Likes(John,Peanuts)")

# Resolve with negated goal
print("\nStep 4:")
print("Likes(John,Peanuts)")
print("~Likes(John,Peanuts)")
print("Therefore: Empty Clause []")

print("\nConclusion:")
print("Empty clause obtained.")
print("Hence, John likes peanuts is PROVED by Resolution.")
print("BHOOMIKA B  1WN24CS063")
```
