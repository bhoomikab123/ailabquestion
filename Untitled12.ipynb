```python
from termcolor import colored

# --------------------------------------------------
# GOAL STATE
# --------------------------------------------------

goal_state = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]


# --------------------------------------------------
# PRINT BOARD
# --------------------------------------------------

def print_board(board):
    print("+---+---+---+")

    for row in range(0, 9, 3):
        row_visual = "|"

        for tile in board[row:row + 3]:
            if tile == 0:
                row_visual += f" {colored(' ', 'cyan')} |"
            else:
                row_visual += f" {colored(str(tile), 'yellow')} |"

        print(row_visual)
        print("+---+---+---+")


# --------------------------------------------------
# POSSIBLE MOVES
# --------------------------------------------------

moves = {
    'U': -3,
    'D': 3,
    'L': -1,
    'R': 1
}


# --------------------------------------------------
# MOVE TILE
# --------------------------------------------------

def move_tile(board, move, blank_pos):
    new_board = board[:]
    new_blank_pos = blank_pos + moves[move]

    new_board[blank_pos], new_board[new_blank_pos] = (
        new_board[new_blank_pos],
        new_board[blank_pos]
    )

    return new_board


# --------------------------------------------------
# DEPTH-LIMITED DFS
# --------------------------------------------------

def depth_limited_search(board, depth_limit, path, visited):
    state = tuple(board)

    # Goal test
    if board == goal_state:
        return path

    # Stop when the depth limit is reached
    if len(path) >= depth_limit:
        return None

    visited.add(state)

    blank_pos = board.index(0)

    for move in moves:

        # Check valid moves
        if move == 'U' and blank_pos < 3:
            continue

        if move == 'D' and blank_pos > 5:
            continue

        if move == 'L' and blank_pos % 3 == 0:
            continue

        if move == 'R' and blank_pos % 3 == 2:
            continue

        # Generate the next state
        new_board = move_tile(board, move, blank_pos)
        new_state = tuple(new_board)

        # Avoid revisiting states in the current search
        if new_state in visited:
            continue

        result = depth_limited_search(
            new_board,
            depth_limit,
            path + [(new_board, move)],
            visited
        )

        if result is not None:
            return result

    visited.remove(state)
    return None




def ids(start_state):

    depth_limit = 0

    while True:

        print(f"\nSearching at depth limit: {depth_limit}")


        visited = set()

        result = depth_limited_search(
            start_state,
            depth_limit,
            [(start_state, None)],
            visited
        )

        if result is not None:
            return result

        depth_limit += 1




def print_solution(solution):

    print("\n========== SOLUTION FOUND ==========\n")

    for depth, (board, move) in enumerate(solution):

        print(
            f"Move: {move} | "
            f"Depth: {depth}"
        )

        print_board(board)




initial_state = [
    2, 8, 3,
    1, 6, 4,
    0, 7, 5
]

print("Initial State:")
print_board(initial_state)

print("\nGoal State:")
print_board(goal_state)

solution = ids(initial_state)

if solution:
    print_solution(solution)
else:
    print(colored("No solution exists.", "red"))

print("\nBhoomika B_1WN24CS063")
```
