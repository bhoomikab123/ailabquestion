```python
import heapq
from termcolor import colored


class PuzzleState:
    def __init__(self, board, parent, move, depth, cost):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth          # g(n)
        self.cost = cost            # f(n) = g(n) + h(n)

    def __lt__(self, other):
        return self.cost < other.cost


# --------------------------------------------------
# GOAL STATE
# --------------------------------------------------

goal_state = [1, 2, 3,
              8, 0, 4,
              7, 6, 5]


# --------------------------------------------------
# PRINT BOARD
# --------------------------------------------------

def print_board(board):
    print("+---+---+---+")

    for row in range(0, 9, 3):
        row_visual = "|"

        for tile in board[row:row + 3]:

            if tile == 0:
                row_visual += f" {colored(' ', 'cyan')} |"
            else:
                row_visual += f" {colored(str(tile), 'yellow')} |"

        print(row_visual)
        print("+---+---+---+")


# --------------------------------------------------
# POSSIBLE MOVES
# --------------------------------------------------

moves = {
    'U': -3,
    'D': 3,
    'L': -1,
    'R': 1
}


# --------------------------------------------------
# HEURISTIC FUNCTION
# Manhattan Distance
# h(n)
# --------------------------------------------------

def heuristic(board):

    distance = 0

    for i in range(9):

        tile = board[i]

        # Do not calculate distance for blank tile
        if tile != 0:

            # Current position
            current_row, current_col = divmod(i, 3)

            # Correct goal position for this tile
            goal_position = goal_state.index(tile)

            goal_row, goal_col = divmod(goal_position, 3)

            # Manhattan distance
            distance += abs(current_row - goal_row) + \
                        abs(current_col - goal_col)

    return distance


# --------------------------------------------------
# MOVE TILE
# --------------------------------------------------

def move_tile(board, move, blank_pos):

    new_board = board[:]

    new_blank_pos = blank_pos + moves[move]

    # Swap blank and tile
    new_board[blank_pos], new_board[new_blank_pos] = \
        new_board[new_blank_pos], new_board[blank_pos]

    return new_board


# --------------------------------------------------
# A* SEARCH
# --------------------------------------------------

def a_star(start_state):

    open_list = []
    closed_list = set()

    # Initial state
    initial_h = heuristic(start_state)

    initial_state = PuzzleState(
        start_state,
        None,
        None,
        0,
        initial_h
    )

    heapq.heappush(open_list, initial_state)

    while open_list:

        # Get state with smallest f(n)
        current_state = heapq.heappop(open_list)

        # Check whether goal is reached
        if current_state.board == goal_state:
            return current_state

        # Add current state to closed list
        closed_list.add(tuple(current_state.board))

        # Find blank position
        blank_pos = current_state.board.index(0)

        # Try all possible moves
        for move in moves:

            # --------------------------------------
            # Check whether move is valid
            # --------------------------------------

            if move == 'U' and blank_pos < 3:
                continue

            if move == 'D' and blank_pos > 5:
                continue

            if move == 'L' and blank_pos % 3 == 0:
                continue

            if move == 'R' and blank_pos % 3 == 2:
                continue

            # --------------------------------------
            # Generate new board
            # --------------------------------------

            new_board = move_tile(
                current_state.board,
                move,
                blank_pos
            )

            # Ignore already visited states
            if tuple(new_board) in closed_list:
                continue

            # --------------------------------------
            # Calculate g(n), h(n), f(n)
            # --------------------------------------

            new_depth = current_state.depth + 1

            h_value = heuristic(new_board)

            f_value = new_depth + h_value

            # Create new state
            new_state = PuzzleState(
                new_board,
                current_state,
                move,
                new_depth,
                f_value
            )

            # Add to priority queue
            heapq.heappush(open_list, new_state)

    return None


# --------------------------------------------------
# PRINT SOLUTION
# --------------------------------------------------

def print_solution(solution):

    path = []

    current = solution

    # Go backwards from goal to start
    while current:

        path.append(current)

        current = current.parent

    # Reverse the path
    path.reverse()

    print("\n========== SOLUTION ==========\n")

    for step in path:

        h_val = heuristic(step.board)

        print(
            f"Move: {step.move} | "
            f"g(n) = {step.depth} | "
            f"h(n) = {h_val} | "
            f"f(n) = {step.cost}"
        )

        print_board(step.board)


# --------------------------------------------------
# MAIN PROGRAM
# --------------------------------------------------

initial_state = [
    2, 8, 3,
    1, 6, 4,
    0, 7, 5
]


print("Initial State:")
print_board(initial_state)

print("\nGoal State:")
print_board(goal_state)


# Run A*
solution = a_star(initial_state)


# --------------------------------------------------
# DISPLAY RESULT
# --------------------------------------------------

if solution:

    print(colored("\nSolution found!", "green"))

    print_solution(solution)

else:

    print(colored("\nNo solution exists.", "red"))


print("\nBhoomika B_1WN24CS063")
```
