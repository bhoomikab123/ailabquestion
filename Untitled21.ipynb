```python

from itertools import product

print("P Q R | Q->P | P->~Q | Q|R | KB")
print("-------------------------------------")

models = []

for P, Q, R in product([False, True], repeat=3):

    s1 = (not Q) or P
    s2 = (not P) or (not Q)
    s3 = Q or R

    KB = s1 and s2 and s3

    print(
        P, Q, R, "|",
        s1, "|", s2, "|", s3, "|", KB
    )

    if KB:
        models.append((P, Q, R))

print("\nModels of KB:", models)

# Check entailments
entails_R = all(R for P, Q, R in models)

entails_R_implies_P = all(
    (not R) or P for P, Q, R in models
)

entails_Q_implies_R = all(
    (not Q) or R for P, Q, R in models
)

print("\nKB entails R:", entails_R)
print("KB entails R -> P:", entails_R_implies_P)
print("KB entails Q -> R:", entails_Q_implies_R)
print("BHOOMIKA B"+" 1WN24CS063")
```
