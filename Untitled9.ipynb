```python
# 8 Puzzle Solver using IDA* (Depth-First Search with Optimal Heuristic)

# Define the target configuration (0 represents the empty blank tile)
GOAL_STATE = (1, 2, 3,
              0, 4, 6,
              7, 5, 8)

# Map values to their target (row, col) coordinates for Manhattan distance calculation
GOAL_COORDS = {val: (i // 3, i % 3) for i, val in enumerate(GOAL_STATE)}

def get_manhattan_distance(state):
    """Calculates the sum of absolute distances of tiles from their goal positions."""
    distance = 0
    for i, val in enumerate(state):
        if val != 0:
            current_row, current_col = i // 3, i % 3
            goal_row, goal_col = GOAL_COORDS[val]
            distance += abs(current_row - goal_row) + abs(current_col - goal_col)
    return distance

def get_neighbors(state):
    """Generates all valid next states and the corresponding move directions."""
    neighbors = []
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    # Define possible directional offsets for moving the blank space
    moves = {
        'UP': (-1, 0),
        'DOWN': (1, 0),
        'LEFT': (0, -1),
        'RIGHT': (0, 1)
    }

    for direction, (dr, dc) in moves.items():
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank_idx = new_row * 3 + new_col
            # Convert to list to execute tile swap
            state_list = list(state)
            state_list[blank_idx], state_list[new_blank_idx] = state_list[new_blank_idx], state_list[blank_idx]
            neighbors.append((tuple(state_list), direction))

    return neighbors

def dfs_search(path, moves_taken, g, f_limit):
    """
    Core Depth-First Search recursive routine restricted by a cost limit (IDA*).
    Returns (True, path, moves_taken) if goal found, or the next minimum exceeded threshold integer.
    """
    current_state = path[-1]
    f = g + get_manhattan_distance(current_state)

    if f > f_limit:
        return f
    if current_state == GOAL_STATE:
        return True

    min_threshold = float('inf')

    for neighbor, direction in get_neighbors(current_state):
        if neighbor not in path:  # Avoid cycles in the current DFS branch
            path.append(neighbor)
            moves_taken.append(direction)

            result = dfs_search(path, moves_taken, g + 1, f_limit)
            if result is True:
                return True
            if result < min_threshold:
                min_threshold = result

            moves_taken.pop()  # Backtrack Move
            path.pop()         # Backtrack DFS State

    return min_threshold

def solve_8_puzzle(start_state):
    """Initializes the IDA* algorithm and increments depth limits progressively."""
    # Convert input list or matrix to a flat tuple for fast hashing/matching
    if isinstance(start_state[0], list):
        start_state = tuple(item for row in start_state for item in row)
    else:
        start_state = tuple(start_state)

    f_limit = get_manhattan_distance(start_state)
    path = [start_state]
    moves_taken = []

    while True:
        result = dfs_search(path, moves_taken, 0, f_limit)
        if result is True:
            return path, moves_taken
        if result == float('inf'):
            return None, None  # Unsolvable puzzle configuration
        f_limit = result

def print_solution(path, moves_taken):
    """Prints the step-by-step matrix sequence, path costs, and direction moves to reach the goal state."""
    if not path:
        print("No solution exists.")
        return

    total_cost = len(path) - 1
    print(f"Optimal solution found in {total_cost} steps!\n")

    for step, state in enumerate(path):
        # Determine the action text
        if step == 0:
            action_str = "Initial State"
        else:
            action_str = f"Move Blank Space {moves_taken[step - 1]}"

        print(f"--- Step {step} | {action_str} | Path Cost (g): {step} ---")
        for i in range(0, 9, 3):
            print(list(state[i:i+3]))
        print()

# --- Execution Example ---
if __name__ == "__main__":
    # 0 represents the blank space.
    initial_puzzle = [[1, 2, 3],
                      [4, 0, 6],
                      [7, 5, 8]]

    solution_path, actions = solve_8_puzzle(initial_puzzle)
    print_solution(solution_path, actions)

    # Custom Signature Output
    print("Bhoomika B")
    print("1WN24CS063")
```
