```python
# 8 Puzzle Solver using IDA*
# IDA* = Iterative Deepening A*
# Heuristic = Manhattan Distance

# ---------------------------------------------------------
# GOAL STATE
# ---------------------------------------------------------

# 0 represents the blank space
GOAL_STATE = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


# ---------------------------------------------------------
# GOAL COORDINATES
# ---------------------------------------------------------

GOAL_COORDS = {
    val: (i // 3, i % 3)
    for i, val in enumerate(GOAL_STATE)
}


# ---------------------------------------------------------
# MANHATTAN DISTANCE
# ---------------------------------------------------------

def get_manhattan_distance(state):

    distance = 0

    for i, val in enumerate(state):

        # Ignore blank
        if val != 0:

            current_row = i // 3
            current_col = i % 3

            goal_row, goal_col = GOAL_COORDS[val]

            distance += (
                abs(current_row - goal_row)
                + abs(current_col - goal_col)
            )

    return distance


# ---------------------------------------------------------
# GENERATE NEIGHBORS
# ---------------------------------------------------------

def get_neighbors(state):

    neighbors = []

    # Find blank position
    blank_idx = state.index(0)

    row = blank_idx // 3
    col = blank_idx % 3

    # Possible blank movements
    moves = {
        "UP": (-1, 0),
        "DOWN": (1, 0),
        "LEFT": (0, -1),
        "RIGHT": (0, 1)
    }

    for direction, (dr, dc) in moves.items():

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank_idx = new_row * 3 + new_col

            state_list = list(state)

            # Swap blank and tile
            state_list[blank_idx], state_list[new_blank_idx] = (
                state_list[new_blank_idx],
                state_list[blank_idx]
            )

            new_state = tuple(state_list)

            neighbors.append(
                (new_state, direction)
            )

    return neighbors


# ---------------------------------------------------------
# IDA* DFS SEARCH
# ---------------------------------------------------------

def dfs_search(path, moves_taken, g, f_limit):

    current_state = path[-1]

    # h(n) = Manhattan distance
    h = get_manhattan_distance(current_state)

    # f(n) = g(n) + h(n)
    f = g + h

    # If cost exceeds limit
    if f > f_limit:
        return f

    # Goal test
    if current_state == GOAL_STATE:
        return True

    min_threshold = float("inf")

    # Generate next states
    for neighbor, direction in get_neighbors(current_state):

        # Avoid cycles
        if neighbor not in path:

            path.append(neighbor)
            moves_taken.append(direction)

            # Recursive DFS
            result = dfs_search(
                path,
                moves_taken,
                g + 1,
                f_limit
            )

            # Goal found
            if result is True:
                return True

            # Find minimum threshold
            if result < min_threshold:
                min_threshold = result

            # Backtracking
            moves_taken.pop()
            path.pop()

    return min_threshold


# ---------------------------------------------------------
# SOLVE PUZZLE
# ---------------------------------------------------------

def solve_8_puzzle(start_state):

    # Convert matrix to tuple
    if isinstance(start_state[0], list):

        start_state = tuple(
            item
            for row in start_state
            for item in row
        )

    else:

        start_state = tuple(start_state)

    # Initial threshold
    f_limit = get_manhattan_distance(start_state)

    # Store path
    path = [start_state]

    # Store actions
    moves_taken = []

    # IDA* loop
    while True:

        result = dfs_search(
            path,
            moves_taken,
            0,
            f_limit
        )

        # Solution found
        if result is True:
            return path, moves_taken

        # No solution
        if result == float("inf"):
            return None, None

        # Increase threshold
        f_limit = result


# ---------------------------------------------------------
# PRINT SOLUTION
# ---------------------------------------------------------

def print_solution(path, moves_taken):

    if not path:

        print("No solution exists.")
        return

    total_cost = len(path) - 1

    print(
        f"Optimal solution found in {total_cost} steps!\n"
    )

    for step, state in enumerate(path):

        if step == 0:

            action_str = "Initial State"

        else:

            action_str = (
                f"Move Blank Space "
                f"{moves_taken[step - 1]}"
            )

        print(
            f"--- Step {step} | "
            f"{action_str} | "
            f"Path Cost (g): {step} ---"
        )

        # Print 3x3 puzzle
        for i in range(0, 9, 3):

            print(
                list(state[i:i + 3])
            )

        print()


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":

    # Initial puzzle
    initial_puzzle = [
        [1, 2, 3],
        [4, 0, 6],
        [7, 5, 8]
    ]

    # Solve
    solution_path, actions = solve_8_puzzle(
        initial_puzzle
    )

    # Print solution
    print_solution(
        solution_path,
        actions
    )

    # Student details
    print("Bhoomika B")
    print("1WN24CS063")
```
